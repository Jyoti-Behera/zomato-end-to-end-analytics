```python
#Country wise Restaurant Distribution
country_count=df['Country Code'].value_counts()
country_count.head()
```

```python
#Visulaization
plt.figure(figsize=(8,5))
sns.countplot(x='Country Code',data=df)
plt.title("Country-wise Restaurant Distribution")

plt.show()
```

```python
#Insights
#India dominates the zomato platform.
```

```python
#City wise Top 10 restaurant location
df['City'].value_counts().head(10)
```

```python
#visualization
plt.figure(figsize=(10,5))
df['City'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Cities with maximum restaurants")

plt.show()
```

```python
#major metro cities have higher restaurant density.
```

```python
#online delivery analysis
df['Has Online delivery'].value_counts()
```

```python
#visulaization
plt.figure(figsize=(6,4))
sns.countplot(x="Has Online delivery",data=df)
plt.title("Online Delivery Availability")

plt.show()
```

```python
# most restaurants donot offer online delivery.
```

```python
#table booking analysis
sns.countplot(x="Has Table booking",data=df)
plt.title("Table Booking Availability")

plt.show()
```

```python
#Table booking is availaible at limited restaurants.
```

```python
#Rating Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['Aggregate rating'],bins=10,kde=True)
plt.title("Distribution of Aggregate Ratings")

plt.show()
```

```python
#Most ratings lies between 3.0 and 4.5
```

```python
# rating test analysis
df['Rating text'].value_counts()
```

```python
# visualisation
sns.countplot(y='Rating text',data=df)
plt.title("Rating Text Distribution")

plt.show()
```

```python
# Average cost of 2 analysis
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Average Cost for two'])
plt.title("Average Cost for Two")

plt.show()
```

```python
# Most restauranys falls under mid price range.
```

```python
# cost vs mid rating restaurants
plt.figure(figsize=(8,5))
sns.scatterplot(x='Average Cost for two',y='Aggregate rating',data=df)
plt.title("Cost vs Rating")

plt.show()

```

```python
# Higher cost doesnot gurantees higher ratings.
```

```python
# Price range vs online delivery
sns.countplot(x='Price range',hue='Has Online delivery',data =df)
plt.title("Price Range vs Online Delivery")

plt.show()
```

```python
# Mid price restaurants are likely to offer delivery.
```

```python
#Votes vs Ratings
plt.figure(figsize=(8,5))
sns.scatterplot(x='Votes',y='Aggregate rating',data=df)
plt.title("Votes vs Rating")

plt.show()
```

```python
# Restaurants with more votings are more stable.
```

```python
# Top 10 popular cuisines
cuisines=df['Cuisines'].str.split(',').explode()
top_cuisines=cuisines.value_counts().head(10)

top_cuisines.plot(kind='bar',figsize=(8,5))
plt.title("Top 10 Popular Cuisines")

plt.show()
```

```python
#Conclusion
#India has highest number of restaurants.
#online delivery is not universal.
#cost doesnot strongly affects ratings.
#user votes improve rating reliablity.
#cuisine type influences popularity.
```
